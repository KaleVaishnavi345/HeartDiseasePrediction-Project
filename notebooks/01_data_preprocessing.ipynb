```python
import pandas as pd

# Load dataset
df = pd.read_csv(r"D:\Project\Heart_disease\HeartDisease_Project\data\heart.csv")

# Preview
print(df.head())

# Check null values
print(df.isnull().sum())

# Fill numeric missing values
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

# Save processed dataset
df.to_csv(r"D:\Project\Heart_disease\HeartDisease_Project\data\processed_data.csv", index=False)

print("✅ Preprocessing completed and file saved")
```
